```python
def bisection(f, a, b, tol=1e-7, max_iter=100):
    """Finds a root of f in [a, b] using the Bisection Method."""
    if f(a) * f(b) > 0:
        raise ValueError("f(a) and f(b) must have opposite signs")

    for i in range(max_iter):
        c = (a + b) / 2
        if abs(b - a) / 2 < tol:
            return c, i + 1
        if f(a) * f(c) < 0:
            b = c
        else:
            a = c
    return c, max_iter

import numpy as np

# (a) x^3 = 9  ->  f(x) = x^3 - 9
f_a = lambda x: x**3 - 9
root_a, iters_a = bisection(f_a, 2, 3)
print(f"(a) root = {root_a:.6f}  (iterations: {iters_a})")

# (b) 3x^3 + x^2 = x + 5  ->  f(x) = 3x^3 + x^2 - x - 5
f_b = lambda x: 3*x**3 + x**2 - x - 5
root_b, iters_b = bisection(f_b, 1, 2)
print(f"(b) root = {root_b:.6f}  (iterations: {iters_b})")

# (c) cos^2(x) + 6 = x  ->  f(x) = cos(x)**2 + 6 - x
f_c = lambda x: np.cos(x)**2 + 6 - x
root_c, iters_c = bisection(f_c, 6, 7)
print(f"(c) root = {root_c:.6f}  (iterations: {iters_c})")
```

```python
import numpy as np

def bisection(f, a, b, tol=1e-9, max_iter=200):
    """Finds a root of f in [a, b] using the Bisection Method."""
    if f(a) * f(b) > 0:
        raise ValueError("f(a) and f(b) must have opposite signs")
    for i in range(max_iter):
        c = (a + b) / 2
        if abs(b - a) / 2 < tol:
            return c, i + 1
        if f(a) * f(c) < 0:
            b = c
        else:
            a = c
    return c, max_iter

# (a) x^5 + x = 1  ->  f(x) = x^5 + x - 1
f_a = lambda x: x**5 + x - 1
root_a, iters_a = bisection(f_a, 0, 1)
print(f"(a) root = {root_a:.8f}  (iterations: {iters_a})")

# (b) sin x = 6x + 5  ->  f(x) = sin(x) - 6*x - 5
f_b = lambda x: np.sin(x) - 6*x - 5
root_b, iters_b = bisection(f_b, -1, 0)
print(f"(b) root = {root_b:.8f}  (iterations: {iters_b})")

# (c) ln x + x^2 = 3  ->  f(x) = np.log(x) + x**2 - 3
f_c = lambda x: np.log(x) + x**2 - 3
root_c, iters_c = bisection(f_c, 1, 2)
print(f"(c) root = {root_c:.8f}  (iterations: {iters_c})")
```

```python
import numpy as np
import matplotlib.pyplot as plt

def bisection(f, a, b, tol=1e-7, max_iter=100):
    if f(a) * f(b) > 0:
        raise ValueError("f(a) and f(b) must have opposite signs")
    for i in range(max_iter):
        c = (a + b) / 2
        if abs(b - a) / 2 < tol:
            return c, i + 1
        if f(a) * f(c) < 0:
            b = c
        else:
            a = c
    return c, max_iter
```

```python
f_a = lambda x: 2*x**3 - 6*x - 1

x = np.linspace(-3, 3, 400)
plt.figure()
plt.plot(x, f_a(x))
plt.axhline(0, color='black', linewidth=0.8)
plt.title("2x^3 - 6x - 1")
plt.grid(True)
plt.show()

intervals_a = [(-2, -1), (-1, 0), (1, 2)]
for (a, b) in intervals_a:
    root, iters = bisection(f_a, a, b)
    print(f"root in [{a},{b}] = {root:.6f}  (iterations: {iters})")
```

```python
f_b = lambda x: np.exp(x - 2) + x**3 - x

x = np.linspace(-3, 3, 400)
plt.figure()
plt.plot(x, f_b(x))
plt.axhline(0, color='black', linewidth=0.8)
plt.title("e^(x-2) + x^3 - x")
plt.grid(True)
plt.show()

intervals_b = [(-2, -1), (0, 0.5), (0.5, 1)]
for (a, b) in intervals_b:
    root, iters = bisection(f_b, a, b)
    print(f"root in [{a},{b}] = {root:.6f}  (iterations: {iters})")
```
